# CNN (Convolutional Neural Network) – "Si Detektif Visual"

CNN adalah ahlinya data yang berbentuk spasial atau matriks grid, seperti gambar atau video.

<b>Analogi</b>: Bayangkan kamu sedang mencari wajah temanmu di sebuah foto kerumunan menggunakan kaca pembesar. Kamu menggeser kaca pembesar itu dari pojok kiri atas, mendeteksi pola garis, warna, bentuk hidung, sampai akhirnya mengenali wajah secara utuh. CNN bekerja persis seperti itu menggunakan filter yang bergeser untuk menangkap fitur-fitur penting.

<b>Contoh Penggunaan</b>
- Sistem Face Recognition di HP.
- Deteksi penyakit lewat foto rontgen/MRI.
- Mobil otonom (self-driving car) untuk mendeteksi rambu lalu lintas.

```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Model CNN sederhana untuk klasifikasi gambar
model_cnn = models.Sequential([
    # Layer Convolutional untuk menangkap fitur gambar (pola, garis)
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(64, 64, 3)),
    # Layer Pooling untuk memperkecil ukuran gambar tanpa kehilangan info penting
    layers.MaxPooling2D((2, 2)),
    
    layers.Flatten(),
    # Dense layer untuk klasifikasi akhir
    layers.Dense(64, activation='relu'),
    layers.Dense(2, activation='softmax') # Contoh: 2 kelas (Kucing vs Anjing)
])
```

# RNN (Recurrent Neural Network) – "Si Memori Jangka Pendek"

RNN didesain untuk data yang berurutan (sequential data) atau time-series, di mana <b>urutan data sangat menentukan arti</b>. Data masa lalu memengaruhi prediksi masa kini.

<b>Analogi</b>: Bayangkan kamu sedang membaca sebuah kalimat. Untuk memahami kata kelima, kamu harus ingat kata kesatu sampai keempat. RNN memiliki loop di dalamnya yang membuat informasi dari kata sebelumnya terus dibawa ke proses berikutnya.

<b>Kelemahan</b>: RNN punya penyakit "pikun" (Vanishing Gradient). Kalau kalimat atau urutan datanya terlalu panjang, dia bakal lupa sama informasi yang ada di paling awal.

<b>Contoh Penggunaan</b>
- Prediksi teks otomatis (Auto-complete / Next-word prediction).
- Prediksi cuaca atau suhu jam berikutnya berdasarkan data beberapa jam lalu.

```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Model RNN sederhana untuk data sekuensial
model_rnn = models.Sequential([
    # SimpleRNN memproses urutan teks/angka
    layers.SimpleRNN(32, input_shape=(10, 1), activation='relu'),
    layers.Dense(1) # Output prediksi berikutnya
])
```

# LSTM (Long Short-Term Memory) – "Si Pembuat Catatan yang Pintar"

LSTM adalah versi upgrade dari RNN. LSTM diciptakan khusus <b>untuk menyembuhkan penyakit pikunnya RNN saat menghadapi urutan data yang sangat panjang</b>.

<b>Analogi</b>: Bayangkan kamu sedang membaca buku novel yang tebal. RNN akan mencoba mengingat setiap kata secara mentah-mentah lalu akhirnya lupa jalan ceritanya. Sementara LSTM bertindak seperti orang pintar: dia membawa buku catatan. LSTM punya komponen bernama gates (gerbang) yang secara otomatis memutuskan: "Poin ini penting, catat dan ingat sampai akhir bab" (Forget & Input Gate), dan "Detail ini tidak penting, coret saja" (Output Gate).

<b>Contoh Penggunaan</b>
- Penerjemah bahasa (seperti Google Translate yang harus paham konteks satu paragraf utuh).
- Analisis sentimen teks panjang (review produk atau artikel berita).
- Prediksi pergerakan harga saham atau deteksi pola penipuan (fraud detection) berbasis histori transaksi yang panjang.

```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Model LSTM untuk menangkap dependensi jangka panjang
model_lstm = models.Sequential([
    # Menggunakan layer LSTM sebagai pengganti RNN standar
    layers.LSTM(64, input_shape=(50, 1), activation='tanh'),
    layers.Dense(32, activation='relu'),
    layers.Dense(1)
])
```

# Summary

- <b>CNN</b> : Untuk Gambar/Spasial (Fokus pada apa yang ada di ruang/posisi tertentu).
- <b>RNN</b> : Untuk Urutan Pendek (Fokus pada apa yang terjadi sesaat sebelum ini).
- <b>LSTM</b>: Untuk Urutan Panjang (Fokus pada konteks cerita dari awal sampai akhir).
